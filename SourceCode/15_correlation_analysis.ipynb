```python
FIX_INVALID_DATA_LINK = "data/3_fix_invalid_data.csv"
import pandas as pd

df = pd.read_csv(FIX_INVALID_DATA_LINK)
print(df.columns)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

TARGET_COLUMN = "y"
numeric_variables = df.select_dtypes(include=["int64","float64"]).columns
corr = df[numeric_variables].corr()
plt.figure(figsize=(10,8))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.show()
```

Check:
- variables highly correlated
- potential multicollinearity
- relationships between economic indicators
