```python
FIX_INVALID_DATA_LINK = "data/3_fix_invalid_data.csv"
import pandas as pd

df = pd.read_csv(FIX_INVALID_DATA_LINK)
print(df.columns)
```

Which customer's traits related to the decision to subcribe?


```python
import seaborn as sns
import matplotlib.pyplot as plt

# get categorical variables 
TARGET_COLUMN = "y"
numerical_variables = df.select_dtypes(include=["object"]).columns
print(numerical_variables)

for col in numerical_variables:
    plt.figure(figsize=(10,6))

    rate = pd.crosstab(
        df[col],
        df[TARGET_COLUMN],
        normalize="index",
    )["yes"] * 100

    rate.sort_values().plot(kind="barh")

    plt.xlabel(f"Subscription rate (%) with {col}")
    plt.show()
```
