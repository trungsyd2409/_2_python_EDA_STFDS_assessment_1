```python
REFORMED_DATA_LINK = "data/1_reformed_data.csv"
import pandas as pd
import os

df = pd.read_csv(REFORMED_DATA_LINK)
TARGET_COLUMN_NAME = "y"
```

Q1: How many rows are there in the dataset?

```python
print(f"There are {len(df)} rows in the dataset.")
```

Q2: What about columns?

```python
print(f"There are {len(df.columns)} columns in the dataset.")
```

Q3: What is the target variable?

```python

print(f"{TARGET_COLUMN_NAME} column is the target variable. It is a categorical variable with {df[TARGET_COLUMN_NAME].nunique()} unique values.")
```

Q4: Which are numeric variables?

```python
numerical_columns = df.select_dtypes(include=['int64', 'float64']).columns
print("numerical columns:", numerical_columns)
```

Q5: Which are categorical variables?

```python
categorical_columns = df.select_dtypes(include=['object']).columns
print("categorical columns:", categorical_columns)
```
