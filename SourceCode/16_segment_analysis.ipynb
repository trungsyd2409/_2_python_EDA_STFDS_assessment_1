```python
FIX_INVALID_DATA_LINK = "data/3_fix_invalid_data.csv"
import pandas as pd

df = pd.read_csv(FIX_INVALID_DATA_LINK)
print(df.columns)
```

```python
df["age_group"] = pd.cut(
    df["age"],
    bins=[0, 25, 35, 45, 55, 65, 100],
    labels=[
        "<25",
        "25-34",
        "35-44",
        "45-54",
        "55-64",
        "65+"
    ]
)
df.groupby("age_group")["y"].apply(
    lambda x: (x == "yes").count()/len(df)
) * 100
```
