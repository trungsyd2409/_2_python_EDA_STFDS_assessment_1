Case 2: unknown values

```python
unknown_counts = (df == "unknown").sum()
print(unknown_counts[unknown_counts > 0])
print("--------------------")
print("total data      ", len(df))
```

```python
for col in ["job", "marital", "education", "default", "housing", "loan"]:
    print("\n-----------", col,"------------")
    print(pd.crosstab(
        df[col],
        df["y"],
        normalize="index"
    ))
```

The "unknown" category was retained rather than imputed or removed because it represents information that was unavailable rather than an erroneous value. Replacing unknown values with the mode could introduce bias and create information that was not actually observed. This is particularly important for the default variable, where 20.88% of observations are unknown. Removing these observations would result in substantial data loss.
