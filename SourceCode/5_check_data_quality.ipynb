```python
REFORMED_DATA_LINK = "data/1_reformed_data.csv"
import pandas as pd
import os

df = pd.read_csv(REFORMED_DATA_LINK)
UNKNOWN_VALUE = "unknown"
```

5.1.Check missing value

```python
for col in df.columns:
    print(f"Column '{col}' has {df[col].nunique()} unique values, and {df[col].isnull().sum()} missing values.")
```

5.2. Check special values like unknown

```python

for col in df.columns:
    if UNKNOWN_VALUE in df[col].unique():
        print(f"Column '{col}' has '{UNKNOWN_VALUE}' value.")
```
