```python
REMOVE_UNKNOWN_DATA_LINK = "data/2_remove_unknown_data.csv"
NULL_PROCESSING_DATA_LINK = "data/3_null_processing_data.csv"
```

```python
import pandas as pd

df = pd.read_csv(REMOVE_UNKNOWN_DATA_LINK)
df.isna().sum()
```

```python
numerical_columns = df.select_dtypes(include=['float64', 'int64']).columns
object_columns = df.select_dtypes(include=['object']).columns
for col in numerical_columns:
    df[col] = df[col].fillna(df[col].median())
for col in object_columns:
    df[col] = df[col].fillna(df[col].mode()[0])
print(df.isna().sum())
print("----------------MEAN----------------")
print(round(df.mean(numeric_only=True), 2))
print("----------------MODE----------------")
for column in df.select_dtypes(include="object").columns:
    modes = df[column].mode()

    print(f"{column}: {modes.tolist()}")
```

```python
import csv
with open(NULL_PROCESSING_DATA_LINK, "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    writer.writerow(df.columns)

    for row in df.itertuples(index=False):
        writer.writerow(row)
```
