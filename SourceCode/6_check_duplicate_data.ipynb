```python
REFORMED_DATA_LINK = "data/1_reformed_data.csv"
DROP_DUPLICATES_DATA_LINK = "data/2_drop_duplicates_data.csv"
import pandas as pd
import os

df = pd.read_csv(REFORMED_DATA_LINK)
```

Load file and show that it has any duplicated data

```python

print(f"Number of duplicate rows: {df.duplicated().sum()}")
print("\nDuplicate rows:")
print(df[df.duplicated()])
```

If it has, decide how to handle duplicate data

Solution: decide to drop all duplicate datas because there is no identifier to distinguish between customers

```python
df.drop_duplicates(inplace=True)
print(f"Number of duplicate rows: {df.duplicated().sum()}")
```

Save file

```python
import csv
with open(DROP_DUPLICATES_DATA_LINK, "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    writer.writerow(df.columns)

    for row in df.itertuples(index=False):
        writer.writerow(row)
```
