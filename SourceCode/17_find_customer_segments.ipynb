```python
FIX_INVALID_DATA_LINK = "data/3_fix_invalid_data.csv"
import pandas as pd

df = pd.read_csv(FIX_INVALID_DATA_LINK)
print(df.columns)
```

```python
segment = (
    df.groupby(["contact", "poutcome"])["y"]
      .apply(lambda x: (x == "yes").mean() * 100)
      .reset_index(name="subscription_rate")
)
segment.sort_values(
    "subscription_rate",
    ascending=False
)
```
