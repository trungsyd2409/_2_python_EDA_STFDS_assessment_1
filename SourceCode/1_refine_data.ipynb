Read file data

```python
RAW_DATA_LINK = "data/0_raw_data.csv"
REFORMED_DATA_LINK = "data/1_reformed_data.csv"
REMOVE_UNKNOWN_DATA_LINK = "data/2_remove_unknown_data.csv"
```

```python
import pandas as pd
import os

data = pd.read_csv(RAW_DATA_LINK, sep=";")
features = list(map(lambda x: x.replace('"', ''), data.columns.tolist()[0].split(";")))
rows = data.iloc[:, 0].str.split(";").tolist()
df = pd.DataFrame(rows, columns=features)
df = df.map(lambda x: x.replace('"', '') if isinstance(x, str) else x)
for column in df.columns:
    converted = pd.to_numeric(df[column], errors="coerce")

    if converted.notna().sum() == df[column].notna().sum():
        df[column] = converted
        
len(df)
```

```python
import csv
with open(REFORMED_DATA_LINK, "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    writer.writerow(df.columns)

    for row in df.itertuples(index=False):
        writer.writerow(row)
```

```python
for col in df.columns:
    print(f"{col}: {df[col].nunique()} unique values, {df[col].isnull().sum()} missing values, {df[col].dtype} data type")

```

```python
for column in df.select_dtypes(include="object").columns:
    print(f"\n--- {column} ---")
    print(df[column].value_counts())
```

```python
import numpy as np

for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].replace('unknown', np.nan)

for col in df.columns:
    print(f"{col}: {df[col].nunique()} unique values, {df[col].isnull().sum()} missing values, {df[col].dtype} data type")

for column in df.select_dtypes(include="object").columns:
    print(f"\n--- {column} ---")
    print(df[column].value_counts())
```

```python
with open(REMOVE_UNKNOWN_DATA_LINK, "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    writer.writerow(df.columns)

    for row in df.itertuples(index=False):
        writer.writerow(row)
```
