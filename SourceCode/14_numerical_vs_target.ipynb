```python
FIX_INVALID_DATA_LINK = "data/3_fix_invalid_data.csv"
import pandas as pd

df = pd.read_csv(FIX_INVALID_DATA_LINK)
print(df.columns)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

TARGET_COLUMN = "y"
numeric_variables = df.select_dtypes(include=["int64","float64"]).columns
print(numeric_variables)

for col in numeric_variables:
    sns.boxplot(
        data=df,
        x=TARGET_COLUMN,
        y=col
    )
    plt.show()
    sns.histplot(
        data=df,
        x=col,
        hue=TARGET_COLUMN,
        kde=True
    )
    plt.show()
```
